```python
import numpy as np
from sklearn.tree import DecisionTreeClassifier

counts = np.array([0, 1, 2, 5, 7, 9])
labels = np.array([0, 0, 0, 1, 1, 1])
thresholds = np.arange(0, 11)

# Broadcasting builds a table: one row per threshold, one column per email.
guesses = (counts[None, :] >= thresholds[:, None]).astype(int)
mistakes = (guesses != labels[None, :]).sum(axis=1)
print("mistakes per threshold:", mistakes.tolist())

winners = thresholds[mistakes == mistakes.min()]
print("winning thresholds:", winners.tolist(), "-> middle:", winners.mean())

# The same job done by scikit-learn: a decision tree allowed ONE question (a "stump").
stump = DecisionTreeClassifier(max_depth=1, random_state=0)
stump.fit(counts.reshape(-1, 1), labels)          # features must be a 2-D table: (rows, columns)
print("stump asks: count <=", stump.tree_.threshold[0])
print("stump predictions for 1, 4, 6, 10:", stump.predict([[1], [4], [6], [10]]).tolist())
```
